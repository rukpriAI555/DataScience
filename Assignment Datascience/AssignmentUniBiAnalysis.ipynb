# 1.10 9)Convert the normal distribution to standard normal distribution for
salary column


```python
# For converting normal distribution to standard normal distribution, Mean should be 0 and Standard Deviation should be 1
```

```python
def stdNBgraph (dataset):
    import seaborn as sns
    mean = dataset.mean()
    std = dataset.std()
    values = [ i for i in dataset]
    z_score = [(j-mean)/std for j in values]
    sns.displot(z_score, kde=True)
    print(sum(z_score)/len(z_score))
```

```python
stdNBgraph(dataset["salary"])
```

# 1.11 10)What is the probability Density Function of the salary range from
700000 to 900000?


```python
# Writing a function to find PDF between 700000 and 900000
```

```python
def get_pdf_probability(dataset, startrange, endrange):
    from matplotlib import pyplot
    from scipy.stats import norm
    import seaborn as sns
    ax = sns.histplot(dataset, kde = True, color = 'Green')
    #ax = sns.histplot(dataset, kde = True, kde_kws = {'color':'blue'}, color = 'Green')
    pyplot.axvline (startrange, color = 'Red')
    pyplot.axvline(endrange, color = 'Red')

    #Generate a sample
    sample = dataset

    #calculate parameters
    sample_mean = sample.mean()
    sample_std = sample.std()
    print('Mean = %3f, Standard Deviation = %3f'%(sample_mean, sample_std))

    #define the distribution
    dist = norm(sample_mean, sample_std)

    #sample probabilities for a range of outcomes
    values = [value for value in range(startrange, endrange)]
        
    probabilities = [dist.pdf(value)for value in values]
    
    prob = sum(probabilities)
    print("PDF between 70000 and 90000".format(startrange, endrange, sum(probabilities)))

    return prob
```

```python
#calling the function
get_pdf_probability(dataset["salary"],700000,900000)
```

# 1.14 13) plot any useful graph and explain it.
standard normal distribution Graph


```python
# convert normal distribution to standard normal distribution
def stdNBgraph(dataset):
    import seaborn as sns 
    mean = dataset.mean() 
    std = dataset.std() 
    values = [] 
    # Calculate z-scores 
    for j in dataset: 
        z_score = (j - mean) / std 
        values.append(z_score)

    # Plot the z-scores 
    sns.histplot(values, kde=True)
    print(sum(values)/len(values))
```

```python
stdNBgraph(dataset["ssc_p"])
```

```python
# Standard Normal Distribution graph above has mean = 0 and standard deviation = 1
# x-axis --- from negative range to positive range - only x-axis will change
# Normal Distribution curve can be seen in the above graph
# Standard normal distribution graph will also have histogram
```
