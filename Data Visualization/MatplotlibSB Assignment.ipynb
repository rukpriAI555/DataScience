# Different Types of plots - Matplotlib

```python
#Line Chart
x= [10, 20, 30, 40, 50]
y= [25, 35, 45, 55, 65]
plt.plot(x, y, marker=0)
plt.title("Line Chart")
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
# Bar Chart
x = ["Thur", "Fri", "Sat", "Sun"]
y = [170, 120, 250, 190]
plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
#Histogram
x = [7, 8, 9, 10, 10, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20, 21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40,
     42, 44, 48, 50]
plt.hist(x, bins=10, color='teal',edgecolor = 'black')
plt.title("Histogram")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatter Plot
x = [10, 15, 20, 25, 30]
y = [12, 18, 25, 28, 35]
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
#Pie Chart
cars = ['AUDI', 'TESLA', 'BMW', 'JAGUAR', 'FORD']
data = [23, 10, 35, 15, 12]
plt.pie(data, labels=cars, autopct = '%1.1f%%')
plt.title("PIE CHART")
plt.show()
```

```python
#Boxplot
data = [[10, 12, 14, 15, 18, 20, 22], [8, 9, 11, 13, 17, 19, 21], [14, 16, 18, 20, 23, 25, 27]]
plt.boxplot(data)
plt.title( "Box Plot")
plt.xlabel( "Groups")
plt.ylabel("Values")
plt.show()
```
