```python
#Replacing the square brackets "[" and "]" with nothing '' in the columns Cuisine, Top Dishes and Features
raw_df["Cuisine"]=raw_df["Cuisine"].str.replace('\\[', '', regex=True)
raw_df["Cuisine"]=raw_df["Cuisine"].str.replace('\\]', '', regex=True)
raw_df["Top Dishes"]=raw_df["Top Dishes"].str.replace('\\[', '', regex=True)
raw_df["Top Dishes"]=raw_df["Top Dishes"].str.replace('\\]', '', regex=True)
raw_df["Features"]=raw_df["Features"].str.replace('\\[', '', regex=True)
raw_df["Features"]=raw_df["Features"].str.replace('\\]', '', regex=True)
raw_df.head()
                                               
```

```python
#Removing Restaurant records that have Dining Rating zero and Delivery Rating zero from the Dataset
useless_data = (raw_df["Dining Rating"]==0.0) & (raw_df["Delivery Rating"]==0.0)
raw_df[useless_data]
```

# Copy raw_df to a new data frame for Exploratory Data Analysis

```python
zomato_chennai_df = raw_df.copy()
zomato_chennai_df
```

Question1) How many restaurants are in Chennai,location wise whose Dining Rate or Delivery Rate is more than 4.5?
Basically, we are looking for high rated restaurants in Chennai, location wise

```python
#Find out restaurants whose dining rate or delivery date is more than 4.5
```

```python
highratedres_df = zomato_chennai_df[(zomato_chennai_df["Dining Rating"]>4.5) | (zomato_chennai_df["Delivery Rating"]>4.5)]
highratedres_df
```

```python
#Using histogram to show the number of restaurants location wise that are high rated
fig = px.histogram(highratedres_df, x="Location", color = "Location", title = "Restaurants by Location")
fig.show()
```

Question 2) How many high rated vegetarian restaurants are in Chennai location wise?

```python
veg_df = highratedres_df[highratedres_df["Features"].str.contains("Vegetarian Only")]
veg_df
```

```python
veg_df.shape
```

```python
#Histogram to show number of high rated veg restaurants location wise
fig = px.histogram(veg_df, x="Location", color="Location", title = "High rated Veg Restaurants")
fig.show()
```

Question 3)Percentage of Restaurants by Delivery Rating => 4.5

```python
highdelrate_df = zomato_chennai_df[zomato_chennai_df["Delivery Rating"]>=4.5]
highdelrate_df
```

```python
#Find number of restaurants for each rating type
countres_delrate_df = highdelrate_df["Delivery Rating"].value_counts().reset_index()
countres_delrate_df
```

```python
fig = px.pie(countres_delrate_df, names = "Delivery Rating", values = "count", color="Delivery Rating", 
             title = "Percentage of Restaurants by Delivery Rating").update_traces(textposition='inside',
                                                                                   textinfo ='percent+label')
fig.show()
```
