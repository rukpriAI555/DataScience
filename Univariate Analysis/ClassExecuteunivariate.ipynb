```python
from Univariate import Univariate
import pandas as pd
```

```python
dataset = pd.read_csv("Placement.csv")
```

```python

# Separate quantitative and qualitative
quan, qual = Univariate.QuanQual(dataset)

print("quan:", quan)
print("qual:", qual)
```

```python
# Descriptive analysis
descriptive = Univariate.univariate(dataset, quan)

# Find outliers
lesser, greater = Univariate.Finding_outliers(
    descriptive,
    quan
)

print("Lesser Outliers:", lesser)
print("Greater Outliers:", greater)
```

```python

# Handle outliers
dataset = Univariate.Handle_outliers(
    dataset,
    descriptive,
    quan
)

# Recalculate after handling
descriptive = Univariate.univariate(
    dataset,
    quan
)

# Check again
lesser, greater = Univariate.Finding_outliers(
    descriptive,
    quan
)

print("After Handling:")
print("Lesser Outliers:", lesser)
print("Greater Outliers:", greater)
```
