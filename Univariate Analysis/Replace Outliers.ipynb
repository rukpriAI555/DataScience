```python
def QuanQual(dataset):
    quan = []
    qual = []
    for columnName in dataset.columns:
        #print(columnName)
        if(dataset[columnName].dtypes=='O'):
            qual.append(columnName)
        else:
            quan.append(columnName)
    return quan, qual
```

```python
quan, qual = QuanQual(dataset)
```

```python
descriptive = pd.DataFrame(index = ["Mean", "Median", "Mode", "Q1:25%", "Q2:50%", "Q3:75%", "99%", "Q4:100%", "IQR", "1.5Rule","Lesser", "Greater", "Min", "Max" ], columns = quan)
```

```python
descriptive
```

```python
for columnName in quan:
    descriptive.loc["Mean",columnName ] = dataset[columnName].mean()
    descriptive.loc["Median", columnName] = dataset[columnName].median()
    descriptive.loc["Mode", columnName]=dataset[columnName].mode()[0]
    descriptive.loc["Q1:25%", columnName] = dataset.describe()[columnName]["25%"]
    descriptive.loc["Q2:50%", columnName] = dataset.describe()[columnName]["50%"]
    descriptive.loc["Q3:75%", columnName]= dataset.describe()[columnName]["75%"]
    descriptive.loc["99%", columnName] = np.percentile(dataset[columnName], 99)
    descriptive.loc["Q4:100%", columnName] = dataset.describe()[columnName]["max"]
    descriptive.loc["IQR", columnName] = descriptive.loc["Q3:75%", columnName] - descriptive.loc["Q1:25%", columnName]
    descriptive.loc["1.5Rule", columnName] = 1.5*descriptive.loc["IQR", columnName]
    descriptive.loc["Lesser", columnName] = descriptive.loc["Q1:25%", columnName] - descriptive.loc["1.5Rule", columnName]
    descriptive.loc["Greater", columnName] = descriptive.loc["Q3:75%", columnName] + descriptive.loc["1.5Rule", columnName]
    descriptive.loc["Min", columnName] = dataset[columnName].min()
    descriptive.loc["Max", columnName] = dataset[columnName].max()
```

```python
descriptive
```

```python
def Finding_outliers(descriptive, quan):
    lesser = []
    greater = []

    for columnName in quan:
        if descriptive[columnName]["Min"] < descriptive[columnName]["Lesser"]:
            lesser.append(columnName)

        if descriptive[columnName]["Max"] > descriptive[columnName]["Greater"]:
            greater.append(columnName)

    return lesser, greater
```

```python
Lesser, Greater = Finding_outliers(descriptive, quan)
print("Lesser:", Lesser)
print("Greater:", Greater)
```

```python
def Handle_outliers(dataset, descriptive, quan):
    Lesser, Greater = Finding_outliers(descriptive, quan)

    for ColumnName in Lesser:
        dataset.loc[
            dataset[ColumnName] < descriptive[ColumnName]["Lesser"],
            ColumnName
        ] = descriptive[ColumnName]["Lesser"]

    for ColumnName in Greater:
        dataset.loc[
            dataset[ColumnName] > descriptive[ColumnName]["Greater"],
            ColumnName
        ] = descriptive[ColumnName]["Greater"]

    return dataset


# Replace outliers
dataset = Handle_outliers(dataset, descriptive, quan)
```

```python
# Create updated descriptive table
descriptive_1 = pd.DataFrame(
    index=[
        "Mean", "Median", "Mode",
        "Q1:25%", "Q2:50%", "Q3:75%", "Q4:99%", "Q5:100%",
        "IQR", "1.5Rule", "Lesser", "Greater", "Min", "Max"
    ],
    columns=quan
)

for ColumnName in quan:

    descriptive_1[ColumnName]["Mean"] = dataset[ColumnName].mean()

    descriptive_1[ColumnName]["Median"] = dataset[ColumnName].median()

    descriptive_1[ColumnName]["Mode"] = dataset[ColumnName].mode()[0]

    descriptive_1[ColumnName]["Q1:25%"] = dataset[ColumnName].quantile(0.25)

    descriptive_1[ColumnName]["Q2:50%"] = dataset[ColumnName].quantile(0.50)

    descriptive_1[ColumnName]["Q3:75%"] = dataset[ColumnName].quantile(0.75)

    descriptive_1[ColumnName]["99%"] = np.percentile(
        dataset[ColumnName], 99
    )

    descriptive_1[ColumnName]["Q4:100%"] = dataset[ColumnName].max()

    # Calculate using NEW descriptive_1 values
    descriptive_1[ColumnName]["IQR"] = (
        descriptive_1[ColumnName]["Q3:75%"]
        - descriptive_1[ColumnName]["Q1:25%"]
    )

    descriptive_1[ColumnName]["1.5Rule"] = (
        1.5 * descriptive_1[ColumnName]["IQR"]
    )

    descriptive_1[ColumnName]["Lesser"] = (
        descriptive_1[ColumnName]["Q1:25%"]
        - descriptive_1[ColumnName]["1.5Rule"]
    )

    descriptive_1[ColumnName]["Greater"] = (
        descriptive_1[ColumnName]["Q3:75%"]
        + descriptive_1[ColumnName]["1.5Rule"]
    )

    descriptive_1[ColumnName]["Min"] = dataset[ColumnName].min()

    descriptive_1[ColumnName]["Max"] = dataset[ColumnName].max()


descriptive_1
```
